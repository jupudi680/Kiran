**pd.get_dummies**

- Get dummies is a way to represent the categorical columns to numeerical
- First it will check what are the labels in a column
- For example case status has two labels
    - Denied and Certified
- So it wil create two new columns based on labels
    - Col1: Case_status_Denied
    - Col2: Case_status_Certified

- case_status    |       Denied        |   Certified
-              |      -              |   -

- Deined       |      1               |    0

- Certified     |    0                |     1

- Certified      |   0                 |    1




**It is also called as One Hot Encoder**

    - It makes new columns are independent each other
    - some of Ml algorithm expect that columns should be **independent** Each other
