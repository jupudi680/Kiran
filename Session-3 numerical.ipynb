**Emperical Rule**

```python
wage_data=visa_df['prevailing_wage']
mean=round(wage_data.mean(),2)
mean
```

```python
std=round(wage_data.std(),2)
std
```

```python
lb=mean-1*std
lb
```

```python
ub=mean+1*std
ub
```

```python
con1=wage_data>lb
con2=wage_data<ub
con=con1&con2
np.sum(con)

```

```python
np.percentile(wage_data,68)
```

```python
np.sum(con)==np.percentile(wage_data,68)
```
